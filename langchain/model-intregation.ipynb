```python
from dotenv import load_dotenv
import os

load_dotenv()

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
from langchain.chat_models import init_chat_model
model = init_chat_model("meta-llama/llama-4-scout-17b-16e-instruct", model_provider="groq")
model
```

```python
res = model.invoke("Hello , how are you?")
res
```

```python
for chunk in model.stream("Write me a 200 words paragraph on how ai is bad") :
    print(chunk.text)
```

```python
response = model.batch([
    "Hi, how are you?",
    "why do lightning strike?",
    "what's in uranus?"
    ],
    # configuration
    config = {
        'max_concurrency' : 5
    }
)
for r in response :
    print(r.text)
```
